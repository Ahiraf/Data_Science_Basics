```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv('/content/loan.csv')
dataset.head(5)
```

```python
dataset.describe()
```

```python
plt.figure(figsize=(15, 5)) # Increase figure size for better visualisation and clarity
sns.boxplot(x = "CoapplicantIncome", data = dataset)
plt.show()
```

```python
dataset.shape
```

# **Outlier Removal Using IQR:**

```python
q1 = dataset["CoapplicantIncome"].quantile(0.25)
q3 = dataset["CoapplicantIncome"].quantile(0.75)

IQR = q3 - q1
IQR
```

```python
min_range = q1 - (1.5 * IQR)
max_range = q3 + (1.5 * IQR)

print(f"minimum range: {min_range}, maximum range: {max_range}")
```

```python
newdataset = dataset[dataset["CoapplicantIncome"] <= max_range]
newdataset.shape
```

```python
sns.boxplot(x = "CoapplicantIncome", data = newdataset)
plt.show()
```

### **Outliers can carry meaningful information, and removing them may impact visualisations and downstream analysis. After eliminating the extreme outlier, only a few minor outliers remain. Given their small number and potential significance, retaining them is a reasonable choice.**

# **Outlier Removal Using Z-score**

06096364-0597-4776-BDBA-219AF929BD74_1_201_a.jpeg

```python
z_score = (dataset["CoapplicantIncome"] - dataset["CoapplicantIncome"].mean()) / (dataset["CoapplicantIncome"].std())
z_score
```

```python
z_score > 3
```

```python
dataset["z_score"] = z_score
dataset[dataset["z_score"] < 3] # For removing the outliers
```
